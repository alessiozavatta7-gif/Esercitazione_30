```python
from abc import ABC, abstractmethod


class Person(ABC):
    def __init__(self, nome, cognome, eta):
        self.nome = nome
        self.cognome = cognome
        self.eta = eta

    @abstractmethod
    def descrizione(self):
        pass

    def nome_completo(self):
        return f"{self.nome} {self.cognome}"


class Studente(Person):
    def __init__(self, nome, cognome, eta, matricola, corso):
        super().__init__(nome, cognome, eta)
        self.matricola = matricola
        self.corso = corso

    def descrizione(self):
        return f"{self.nome_completo()} è uno studente di {self.corso} (matricola {self.matricola})"


class Dipendente(Person):
    def __init__(self, nome, cognome, eta, stipendio):
        super().__init__(nome, cognome, eta)
        self.stipendio = stipendio

    def descrizione(self):
        return f"{self.nome_completo()} è un dipendente con stipendio {self.stipendio} €"


class Docente(Dipendente):
    def __init__(self, nome, cognome, eta, stipendio, materia):
        super().__init__(nome, cognome, eta, stipendio)
        self.materia = materia

    def descrizione(self):
        return f"{self.nome_completo()} è un docente di {self.materia}"


class Freelance(Dipendente):
    def __init__(self, nome, cognome, eta, stipendio, partita_iva, tariffa_oraria):
        super().__init__(nome, cognome, eta, stipendio)
        self.partita_iva = partita_iva
        self.tariffa_oraria = tariffa_oraria

    def descrizione(self):
        return f"{self.nome_completo()} è un freelance, {self.tariffa_oraria} €/ora (P.IVA {self.partita_iva})"
```

```python
studente = Studente("Anna", "Rossi", 21, "S12345", "Informatica")
docente = Docente("Luca", "Bianchi", 45, 2200, "Matematica")
freelance = Freelance("Marta", "Verdi", 33, 0, "IT01234567890", 50)

for persona in [studente, docente, freelance]:
    print(persona.descrizione())
```
